```python
aripi = [181, 186, 195, 193, 190, 211]   # lungimea aripii, în mm, la șase pinguini

print(aripi[0])     # primul element (poziția 0)
print(aripi[-1])    # ultimul element
print(aripi[1:3])   # pozițiile 1 și 2; poziția 3 nu intră
print("Câte valori:", len(aripi), "| suma:", sum(aripi), "| cea mai mare:", max(aripi), "| cea mai mică:", min(aripi))
```

## 4. Tabele cu pandas

Biblioteca pandas lucrează cu tabele numite DataFrame: fiecare rând este un exemplu (aici, un pinguin), fiecare coloană o trăsătură. Celula de mai jos citește tabelul direct de pe internet.

| Coloana | Ce conține |
|---|---|
| `species` | specia: Adelie, Chinstrap sau Gentoo |
| `island` | insula: Biscoe, Dream sau Torgersen |
| `bill_length_mm`, `bill_depth_mm` | lungimea și înălțimea ciocului, în mm |
| `flipper_length_mm` | lungimea aripii, în mm |
| `body_mass_g` | masa, în grame |

```python
URL = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
if os.path.exists("penguins.csv"):
    pinguini = pd.read_csv("penguins.csv")   # fișierul încărcat de tine în Colab
else:
    pinguini = pd.read_csv(URL)

print("Rânduri și coloane:", pinguini.shape)
pinguini.head()   # primele 5 rânduri
```

Câteva întrebări pe care i le pui oricărui tabel nou: câte exemple am din fiecare categorie, unde lipsesc valori și cum arată numerele.

```python
print(pinguini["species"].value_counts(), "\n")   # câți pinguini din fiecare specie
print(pinguini.isna().sum())                       # câte valori lipsesc în fiecare coloană
```

```python
pinguini.describe().round(1)   # rezumatul coloanelor numerice: număr, medie, abatere, minim, cuartile, maxim
```

Trei operații pe care le vei folosi mereu:

1. **o coloană**: `pinguini["body_mass_g"]`, cu metode precum `.mean()` (media) sau `.max()`;
2. **filtrarea rândurilor**: `pinguini[pinguini["species"] == "Gentoo"]` păstrează doar rândurile care îndeplinesc condiția (`==` înseamnă „este egal cu”);
3. **gruparea**: `pinguini.groupby("species")["body_mass_g"].mean()` calculează media pe fiecare specie.

```python
print("Masa medie, toți pinguinii:", round(pinguini["body_mass_g"].mean(), 1), "g")

gentoo = pinguini[pinguini["species"] == "Gentoo"]
print("Pinguini Gentoo:", len(gentoo))

pinguini["masa_kg"] = pinguini["body_mass_g"] / 1000   # o coloană nouă, calculată din alta
print(pinguini.groupby("species")["masa_kg"].mean().round(2))
```

## 5. Grafice

Un grafic arată dintr-o privire ce ascund tabelele. Biblioteca matplotlib desenează graficul pe o „figură” (`fig`) cu un sistem de axe (`ax`); fiecare grafic primește titluri pe axe, cu unitatea de măsură.

**Histograma** arată cum se distribuie o singură coloană: câți pinguini au aripa între 180 și 185 mm, între 185 și 190 mm și așa mai departe.

```python
fig, ax = plt.subplots(figsize=(7, 3.5))
ax.hist(pinguini["flipper_length_mm"].dropna(), bins=20, color="#4E6375")   # dropna() scoate valorile lipsă
ax.set_xlabel("Lungimea aripii (mm)")
ax.set_ylabel("Număr de pinguini")
ax.set_title("Cum se distribuie lungimea aripii")
ax.spines[["top", "right"]].set_visible(False)
plt.show()
```

**Graficul cu puncte** pune două coloane una lângă alta: fiecare punct este un pinguin, colorat după specie.

```python
CULORI = {"Adelie": "#4E6375", "Chinstrap": "#9B1B30", "Gentoo": "#161616"}

fig, ax = plt.subplots(figsize=(7, 4.5))
for specie, grup in pinguini.groupby("species"):
    ax.scatter(grup["flipper_length_mm"], grup["body_mass_g"], s=14, color=CULORI[specie], label=specie)
ax.set_xlabel("Lungimea aripii (mm)")
ax.set_ylabel("Masa (g)")
ax.set_title("Aripa și masa, pe specii")
ax.legend(frameon=False)
ax.spines[["top", "right"]].set_visible(False)
plt.show()
```
