## 2. Produsul scalar

Produsul scalar a doi vectori de aceeași lungime: înmulțești element cu element și aduni rezultatele. Este o sumă ponderată.

Exemplul din cursul nostru: nota finală la IPAP se calculează din cinci componente, cu ponderile 30% (portofoliul de laborator), 30% (proiectul), 10% (testele), 20% (proba practică) și 10% (chestionarul). Operatorul `@` calculează produsul scalar.

```python
ponderi = torch.tensor([0.30, 0.30, 0.10, 0.20, 0.10])   # portofoliu, proiect, teste, probă practică, chestionar
note = torch.tensor([8., 9., 10., 7., 6.])                 # notele unui student, în aceeași ordine

print(ponderi * note)           # element cu element
print((ponderi * note).sum())   # suma ponderată
print(ponderi @ note)           # același rezultat, cu operatorul @
```

Un neuron face exact asta: înmulțește intrările cu ponderile lui, adună și mai adaugă un număr, termenul liber. Diferența față de nota finală: ponderile neuronului nu le alege nimeni de mână, ci se învață din date (secțiunea 5).

## 4. Derivata ca pantă

Derivata unei funcții într-un punct este panta: cât de repede crește funcția dacă mergi puțin spre dreapta. O poți calcula direct, cu un pas mic `h`: `(f(x + h) - f(x)) / h`.

Folosim funcția `f(x) = (x - 3)**2 + 1`, care are cea mai mică valoare în x = 3. Imaginează-ți că x este un parametru al modelului, iar f este eroarea lui.

```python
def f(x):
    return (x - 3) ** 2 + 1


def panta_aprox(f, x, h=0.001):
    """Panta lui f în x, aproximată cu un pas mic h."""
    return (f(x + h) - f(x)) / h


for x in [0, 2, 3, 4, 6]:
    print(f"x = {x}:  f(x) = {f(x)},  panta ≈ {panta_aprox(f, x):.2f}")
```

```python
xs = torch.linspace(-1, 7, 200)
x0 = 5
tangenta = f(x0) + panta_aprox(f, x0) * (xs - x0)   # dreapta care are panta funcției în x0

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(xs, f(xs), color="#161616", label="f(x)")
ax.plot(xs, tangenta, color="#9B1B30", linestyle="--", label=f"panta în x = {x0}")
ax.scatter([x0], [f(x0)], color="#9B1B30", zorder=3)
ax.set_ylim(0, 20)
ax.set_xlabel("x (parametrul)")
ax.set_ylabel("f(x) (eroarea)")
ax.legend(frameon=False)
ax.spines[["top", "right"]].set_visible(False)
plt.show()
```

Semnul pantei spune încotro crește funcția. Panta pozitivă: f crește spre dreapta, deci ca f să scadă mergi spre stânga. Panta negativă: invers. Mărimea pantei spune cât de abrupt crește. Pentru această funcție, panta exactă este `2 * (x - 3)`, după regula de derivare; aproximarea dă practic același număr.

```python
def derivata(x):
    """Panta exactă a lui f în x."""
    return 2 * (x - 3)

print(derivata(5), panta_aprox(f, 5))
```

## 6. Gradientul calculat de PyTorch

La o rețea cu milioane de parametri, nimeni nu scrie derivatele de mână. PyTorch le calculează singur: marchezi tensorul cu `requires_grad=True`, calculezi eroarea, apoi apelezi `.backward()`; panta ajunge în `.grad`. Este exact ce a făcut `pierdere.backward()` în caietul L2.

```python
x = torch.tensor(8.0, requires_grad=True)
y = (x - 3) ** 2 + 1
y.backward()
print("Panta calculată de PyTorch:", x.grad.item())
print("Panta din formulă, 2 * (x - 3):", 2 * (8.0 - 3))
```
